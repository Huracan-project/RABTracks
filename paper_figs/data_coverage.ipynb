```python
from eib import *
sns.set_context("paper")
```

```python
# Loading data
tracks = {}
TRACK_flist = glob("../RA_tracks/TRACK-*.pkl")
SyCLoPS_flist = glob("../RA_tracks/SyCLoPS-*.pkl")
for fname in tqdm(TRACK_flist + SyCLoPS_flist):
    name = fname.split("/")[-1].split('.')[0]
    with open(fname, "rb") as f:
        tracks[name] = pkl.load(f)
```

```python
fig = plt.figure(figsize = cm2inch([9,9]))
# Figure 
for i, s in enumerate(tracks):
    Ymin, Ymax = tracks[s].time.dt.year.min().values, tracks[s].time.dt.year.max().values
    plt.plot([Ymin, Ymax], [i, i], marker = 'o', color = PALETTE[s])
    plt.text(Ymin, i+0.2, Ymin, horizontalalignment='center',)
    plt.text(Ymax, i+0.2, Ymax, horizontalalignment='center',)
    # Additionnal turning points
    if s.endswith("ERA5"):
        plt.scatter([1979], [i], marker = "D", color = PALETTE[s])
    if s.endswith("ECMWF-OP-AN"):
        plt.scatter([2010], [i], marker = "D", color = PALETTE[s])
        plt.scatter([2016], [i], marker = "D", color = PALETTE[s])
    
# IBTrACS
plt.plot([1900, 2026], [i+1, i+1], marker = 'o', color = PALETTE["IBTrACS"])
plt.text(2026, i+1.2, 2026, horizontalalignment='center',)

plt.yticks(np.arange(len(tracks)+1), list(tracks.keys())+["IBTrACS"])
plt.xticks([1950, 1975, 2000, 2025])
plt.ylim(-0.5, len(tracks)+1)
plt.xlim(1935,2027)
sns.despine()
plt.grid(axis = 'x')

plt.savefig("data_coverage.png", bbox_inches = "tight")
```
